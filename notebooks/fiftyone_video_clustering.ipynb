```python
from pathlib import Path
import random
import shutil

import numpy as np
import fiftyone as fo
import fiftyone.zoo as foz
import fiftyone.brain as fob
```

```python
NUM_TO_COPY = 100

PROJECT_NAME = "arsenal_mancity_split_video"
DATASET_DIR = Path("../data/arsenal_mancity_split_video")

SAMPLED_DIR = DATASET_DIR.parent / Path(DATASET_DIR.name + f"_{NUM_TO_COPY}")
SAMLED_NAME = PROJECT_NAME + f"_{NUM_TO_COPY}"
```

```python

SAMPLED_DIR.mkdir(parents=True, exist_ok=True)

all_files = list(DATASET_DIR.glob("*.mp4"))
idx = np.round(np.linspace(0, len(all_files) - 1, NUM_TO_COPY)).astype(int)
files_to_copy = [all_files[i] for i in idx]

for files_to_copy in files_to_copy:
    if not (SAMPLED_DIR / files_to_copy.name).exists():
        shutil.copy(files_to_copy, SAMPLED_DIR / files_to_copy.name)
```

```python
# dataset = fo.Dataset.from_videos_dir(DATASET_DIR, name= PROJECT_NAME,overwrite=True, persistent=True, progress=True)
dataset = fo.Dataset.from_images_dir(SAMPLED_DIR, name= SAMLED_NAME,overwrite=True, persistent=True, progress=True)
```

```python
session = fo.launch_app(dataset, port=5151)
```

```python
frames = dataset.to_frames(sample_frames=True, max_fps=1)
```
