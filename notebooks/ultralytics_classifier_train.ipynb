```python

from roboflow import Roboflow
import os

rf = Roboflow(api_key=os.environ.get("ROBOFLOW_API_KEY"))
project = rf.workspace("egroeg121").project("footy-track-broadcast-frame")

DATASET_VERSION = 8

version = project.version(DATASET_VERSION)
dataset = version.download(model_format="folder", location=f"../data/roboflow_dataset_{DATASET_VERSION}")
```

```python
print(f"Dataset location: {dataset.location}")

import os
print("Dataset structure:")
for root, dirs, files in os.walk(dataset.location):
    level = root.replace(dataset.location, '').count(os.sep)
    indent = ' ' * 2 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = ' ' * 2 * (level + 1)
    for file in files[:5]:  # show first 5 files
        print(f"{subindent}{file}")
    if len(files) > 5:
        print(f"{subindent}... and {len(files) - 5} more files")
```

```python
from ultralytics import YOLO

# Load a pretrained classification model
model = YOLO("yolo11n-cls.pt")

# Train the model
# Assuming the dataset is in the correct format with train/ and val/ folders
results = model.train(
    data=dataset.location,
    epochs=50,
    imgsz=224,
    freeze=9,
    cache=True,
    augment=True,
    plots=True,
    # device="mps",
)
```
